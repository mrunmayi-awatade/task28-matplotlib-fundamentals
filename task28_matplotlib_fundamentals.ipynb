```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = {"Month": ["Jan", "Feb", "Mar", "Apr", "May"], "Sales": [100, 150, 120, 180, 200]}
df = pd.DataFrame(data)
print(df)
```

```python
plt.plot(df["Month"], df["Sales"])
plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.show()
```

```python
plt.bar(df["Month"], df["Sales"])
plt.title("Monthly Sales Bar Chart")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.show()
```

```python
plt.scatter(df["Month"], df["Sales"])
plt.title("Monthly Sales Scatter Plot")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.show()
```

```python
plt.hist(df["Sales"], bins=5)
plt.title("Sales Distribution")
plt.xlabel("Sales")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.plot(df["Month"], df["Sales"], color="green", marker="o")
plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.show()
```

```python
plt.plot(df["Month"], df["Sales"], marker="o")
plt.title("Monthly Sales with Grid")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.grid(True)
plt.show()
```

```python
plt.plot(df["Month"], df["Sales"], label="Sales", marker="o")
plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.legend()
plt.show()
```

```python
plt.plot(df["Month"], df["Sales"], marker="o")
plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.savefig("monthly_sales.png")
plt.show()
```

```python
plt.figure(figsize=(10, 5))
plt.plot(df["Month"], df["Sales"], label="Line Chart", marker="o")
plt.bar(df["Month"], df["Sales"], alpha=0.5, label="Bar Chart")
plt.title("Monthly Sales Comparison")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.legend()
plt.show()
```

```python
plt.style.use("ggplot")
plt.plot(df["Month"], df["Sales"], marker="o")
plt.title("Monthly Sales with Style")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.show()
```
